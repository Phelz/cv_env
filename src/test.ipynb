```python
import matplotlib.pyplot as plt
```

```python
def f(x):
    return x * 6.6 *30/ (365*100)
```

```python
# f(9933.51)

initial_amount = 9933.51 + 300
f(initial_amount)

same_pay = 200 - f(initial_amount)

interest_arr = []
amount_arr = []

for i in range(60):
    interest = f(initial_amount)
    same_pay = 200 - interest
    initial_amount = initial_amount - same_pay
    
    interest_arr.append(interest)
    amount_arr.append(initial_amount)
    
#     i_prior = f(i_prior)
#     print(i_prior)
```
